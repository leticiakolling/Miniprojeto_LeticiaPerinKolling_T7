# **Tratamento dos Dados**

```python
# Converter 'DATA' para formato datetime
df['DATA'] = pd.to_datetime(df['DATA'], format='%d/%m/%Y')

print("Tipo de dado da coluna 'DATA' após a conversão:")
print(df['DATA'].dtype)
```

```python
#Verificando valores nulos no DataFrame
print(df.isnull().sum())
print('\n')
```

```python
# Remover as colunas completamente vazias
colunas_para_remover = ['Unnamed: 10', 'Unnamed: 11', 'Unnamed: 12', 'Unnamed: 13']
df.drop(columns=colunas_para_remover, inplace=True)

# Exibir as primeiras linhas após a remoção
df.head()
```

```python
# Verificar linhas duplicadas no DataFrame
duplicados = df.duplicated().sum()
print(f'{duplicados} linhas duplicadas encontradas.')
```

```python
# Remover as linhas duplicadas
df.drop_duplicates(inplace=True)
print(f'DataFrame agora tem {df.shape[0]} linhas após a remoção de duplicatas.')
```

```python
# Filtrar apenas as linhas com '#N/D' na coluna PR_CAT
linhas_nd = df[df['PR_CAT'] == '#N/D']

# Exibir a quantidade total encontrada
print(f"Total de linhas com '#N/D' na coluna PR_CAT: {len(linhas_nd)}")

# Exibir as primeiras linhas desse filtro
display(linhas_nd.head())
```

```python
# Substituir '#N/D' por 'SEM CATEGORIA' na coluna PR_CAT
df['PR_CAT'] = df['PR_CAT'].replace('#N/D', 'SEM CATEGORIA')

# Verificar se a substituição foi concluída com sucesso
print("Quantidade de '#N/D' restante:", (df['PR_CAT'] == '#N/D').sum())
print("Quantidade de 'SEM CATEGORIA' na coluna:", (df['PR_CAT'] == 'SEM CATEGORIA').sum())
```

```python
# Substituir '#N/D' por 'NÃO IDENTIFICADO' na coluna PR_NOME
df['PR_NOME'] = df['PR_NOME'].replace('#N/D', 'NÃO IDENTIFICADO')

# Verificar se a substituição foi concluída com sucesso
print("Quantidade de '#N/D' restante:", (df['PR_NOME'] == '#N/D').sum())
print("Quantidade de 'NÃO IDENTIFICADO' na coluna:", (df['PR_NOME'] == 'NÃO IDENTIFICADO').sum())
```

```python
# Estatísticas descritivas básicas para a coluna CL_FHL
estatisticas = df['CL_FHL'].describe()
mediana = df['CL_FHL'].median()
moda = df['CL_FHL'].mode()[0]

# Exibindo os resultados formatados
print("--- Estatísticas Descritivas para CL_FHL ---")
print(f"Contagem (N):     {estatisticas['count']:.0f}")
print(f"Média:            {estatisticas['mean']:.4f}")
print(f"Mediana:          {mediana:.4f}")
print(f"Desvio Padrão:    {estatisticas['std']:.4f}")
print(f"Moda:             {moda:.4f}")
print(f"Mínimo:           {estatisticas['min']:.4f}")
print(f"Q1 (25%):         {estatisticas['25%']:.4f}")
print(f"Q2 (50%):         {estatisticas['50%']:.4f}")
print(f"Q3 (75%):         {estatisticas['75%']:.4f}")
print(f"Máximo:           {estatisticas['max']:.4f}")
```

```python
# Agrupar por categoria de produto (PR_CAT) e contar a quantidade de vendas
vendas_por_categoria = df.groupby('PR_CAT').size().reset_index(name='QUANTIDADE_VENDAS')

# Calcular o percentual que cada categoria representa sobre o total
total_vendas = vendas_por_categoria['QUANTIDADE_VENDAS'].sum()
vendas_por_categoria['PERCENTUAL'] = (vendas_por_categoria['QUANTIDADE_VENDAS'] / total_vendas) * 100

# Ordenar os resultados do maior para o menor com base na quantidade de vendas
vendas_por_categoria = vendas_por_categoria.sort_values(by='QUANTIDADE_VENDAS', ascending=False)

# Formatar a coluna de percentual como string com duas casas decimais e o símbolo '%'
vendas_por_categoria['PERCENTUAL'] = vendas_por_categoria['PERCENTUAL'].map('{:.2f}%'.format)

# Exibir o resultado
display(vendas_por_categoria)
```

```python
# Agrupar por segmento de cliente (CL_SEG) e contar a quantidade de vendas
vendas_por_segto_cliente = df.groupby('CL_SEG').size().reset_index(name='QUANTIDADE_VENDAS')

# Calcular o percentual que cada segmento representa sobre o total
total_vendas_seg = vendas_por_segto_cliente['QUANTIDADE_VENDAS'].sum()
vendas_por_segto_cliente['PERCENTUAL'] = (vendas_por_segto_cliente['QUANTIDADE_VENDAS'] / total_vendas_seg) * 100

# Ordenar os resultados por segmento de forma crescente
vendas_por_segto_cliente = vendas_por_segto_cliente.sort_values(by='CL_SEG', ascending=True)

# Formatar a coluna de percentual como string com duas casas decimais e o símbolo '%'
vendas_por_segto_cliente['PERCENTUAL'] = vendas_por_segto_cliente['PERCENTUAL'].map('{:.2f}%'.format)

# Exibir o resultado
display(vendas_por_segto_cliente)
```

```python
# Criar uma cópia limpa do DataFrame tratado
df_limpo = df.copy()

# Exibir informações básicas do DataFrame limpo para validação
print("DataFrame 'df_limpo' criado com sucesso!")
print(f"Dimensões: {df_limpo.shape[0]} linhas e {df_limpo.shape[1]} colunas.")
display(df_limpo.head())
```
